```python
import pandas as pd
import numpy as np
from scipy import stats
```

<h1 style="font-size:40px;"> 1.Load Data</h1>

```python
df=pd.read_csv(r"C:\Users\Admin\OneDrive\Desktop\Mathematicas\health_records_300.csv")
print("Dataset Shape:", df.shape)
print(df.head())
```

<h1 style="font-size:40px;"> 2. CONFIDENCE INTERVAL</h1>

```python
print("\n--- 95% CONFIDENCE INTERVAL ---")

for column in ['age', 'weight', 'bmi']:

    data = df[column].dropna()

    mean = data.mean()
    se = stats.sem(data)

    ci = stats.t.interval(
        0.95,
        len(data) - 1,
        loc=mean,
        scale=se
    )

    print(column)
    print("Mean:", round(mean, 2))
    print("95% CI:", tuple(round(x, 2) for x in ci))
```

<h1 style="font-size:40px;">3. T-TEST</h1>

```python
print("\n--- T-TEST ---")

smoker = df[
    df['smoking_status'] == 'Smoker'
]['glucose_level']

non_smoker = df[
    df['smoking_status'] == 'Non-Smoker'
]['glucose_level']

t_stat, p_value = stats.ttest_ind(
    smoker,
    non_smoker,
    equal_var=False
)

print("Smoker Mean:", round(smoker.mean(), 2))
print("Non-Smoker Mean:", round(non_smoker.mean(), 2))
print("t-statistic:", round(t_stat, 3))
print("p-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

<h1 style="font-size:40px;">4. CHI-SQUARE TEST</h1>

```python
print("\n--- CHI-SQUARE TEST ---")

table = pd.crosstab(
    df['smoking_status'],
    df['diabetes']
)

print(table)

chi2, p_value, dof, expected = stats.chi2_contingency(table)

print("Chi-square:", round(chi2, 3))
print("p-value:", round(p_value, 4))
print("Degrees of Freedom:", dof)

critical_value = stats.chi2.ppf(0.95, dof)

print("Critical Value:", round(critical_value, 3))

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

<h1 style="font-size:40px;">5. ANOVA</h1>

```python
print("\n--- ANOVA TEST ---")

groups = [
    group['diabetes'].astype(int)
    for name, group in df.groupby('age_group')
]

f_stat, p_value = stats.f_oneway(*groups)

print("F-statistic:", round(f_stat, 3))
print("p-value:", p_value)

critical_f = stats.f.ppf(
    0.95,
    4,
    295
)

print("Critical F:", round(critical_f, 3))

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```

<h1 style="font-size:40px;"> 6. COVARIANCE</h1>

```python
print("\n--- COVARIANCE ---")

covariance = df['age'].cov(df['bmi'])

print("Age-BMI Covariance:", round(covariance, 3))
```

<h1 style="font-size:40px;"> 7. CORRELATION</h1>

```python
print("\n--- CORRELATION ---")

r, p_value = stats.pearsonr(
    df['age'],
    df['bmi']
)

print("Correlation:", round(r, 3))
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to Reject H0")
```
